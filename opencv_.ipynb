```python
img_ = cv2.imread(r'Data/cats.jpg')

imgrgb = cv2.cvtColor(img_,cv2.COLOR_BGR2RGB)
imggray = cv2.cvtColor(imgrgb,cv2.COLOR_BGR2GRAY)
imgblur = cv2.GaussianBlur(imggray,(7,7),0)
canny_rgb = cv2.Canny(imgrgb,75,150)
canny_blur = cv2.Canny(imgblur,75,100)

plt.subplot(2,2,1)
plt.imshow(imgrgb,cmap='gray')

plt.subplot(2,2,2)
plt.imshow(canny_blur,cmap='gray')
```

```python
blank = np.zeros((500, 500,3), dtype='uint8')

plt.imshow(blank)
```

```python
blank[250:350,250:350] = 0,0,255
```

```python
plt.imshow(blank)
```

```python
cv2.imshow('image', blank)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.rectangle(blank,(0,0),(250,250),(0,255,0),thickness=2)

plt.imshow(blank,cmap='gray')
plt.title('Edge Image')
plt.axis('off')
plt.show()
```

```python
cv2.circle(blank,(225,225),100,(255,0,0),thickness=-1)

plt.imshow(blank,cmap='gray')
plt.title('Edge Image')
plt.axis('off')
plt.show()
```

```python
cv2.putText(blank,'OpenCV',(225,225),cv2.FONT_HERSHEY_SIMPLEX,1,(255,255,255),2)

plt.imshow(blank)
plt.title('Edge Image')
plt.axis('off')
plt.show()
```

```python
captures = cv2.VideoCapture(r'Data/dog.mp4')
```

```python
ret, frame = captures.read()
```

```python
while True:
    ret, frame = captures.read()


    if not ret:
        break


    cv2.imshow('Video', frame)


    if cv2.waitKey(20) & 0xFF == ord('q'):
        break

captures.release()
cv2.destroyAllWindows()
```

```python
while True:
    ret, frame = captures.read()


    if not ret:
        break

    img_gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(img_gray,(7,7),0)
    canny = cv2.Canny(blur,75,150)
    cv2.imshow('Video', canny)

    if cv2.waitKey(20) & 0xFF == ord('q'):
        break

captures.release()
cv2.destroyAllWindows()
```
